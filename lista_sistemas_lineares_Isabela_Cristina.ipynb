## Exercício 2 — Eliminação de Gauss (manual)

Matriz aumentada inicial:
$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
-3 & -1 & 2 & -11 \\
-2 & 1 & 2 & -3
\end{array}\right]
$$

**Etapa 1:** zerar a coluna 1 abaixo do pivô $a_{11} = 2$

- $L_2 \leftarrow L_2 + \frac{3}{2}L_1$
- $L_3 \leftarrow L_3 + L_1$

$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & 1/2 & 1/2 & 1 \\
0 & 2 & 1 & 5
\end{array}\right]
$$

**Etapa 2:** zerar a coluna 2 abaixo do pivô $a_{22} = 1/2$

- $L_3 \leftarrow L_3 - 4L_2$

$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & 1/2 & 1/2 & 1 \\
0 & 0 & -1 & 1
\end{array}\right]
$$

**Substituição regressiva:**

- $-z = 1 \Rightarrow z = -1$
- $\frac{1}{2}y + \frac{1}{2}(-1) = 1 \Rightarrow y - 1 = 2 \Rightarrow y = 3$
- $2x + 3 - (-1) = 8 \Rightarrow 2x = 4 \Rightarrow x = 2$

**Solução:** $x = 2,\ y = 3,\ z = -1$

Verificação na 2ª equação: $-3(2) - 3 + 2(-1) = -11$ ✓

# Parte 2 — Propriedades e interpretação

## Exercício 3 — Número de soluções

**a)** $\begin{cases} x + y = 3 \\ 2x + 2y = 6 \end{cases}$

$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 6 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 0 \end{array}\right]
$$

A segunda linha virou $0 = 0$, então sobra só uma equação pra duas incógnitas.
$\text{posto}(A) = \text{posto}([A|b]) = 1 < 2$ → **infinitas soluções**: $x = 3 - t,\ y = t,\ t \in \mathbb{R}$.

**b)** $\begin{cases} x + y = 3 \\ 2x + 2y = 7 \end{cases}$

$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 7 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 1 \end{array}\right]
$$

A última linha dá $0 = 1$, que é impossível. $\text{posto}(A) = 1 \neq \text{posto}([A|b]) = 2$ → **nenhuma solução** (retas paralelas).

**c)** $\begin{cases} x + y = 3 \\ 2x - y = 0 \end{cases}$

$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & -1 & 0 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & -3 & -6 \end{array}\right]
$$

$-3y = -6 \Rightarrow y = 2$ e $x = 3 - 2 = 1$. $\text{posto}(A) = \text{posto}([A|b]) = 2$ = nº de incógnitas → **solução única** $(1, 2)$.

```python
# Exercicio 3 - conferindo pelo posto (rank) das matrizes
# Teorema de Rouche-Capelli:
#   posto(A) != posto([A|b])            -> sem solucao
#   posto(A) == posto([A|b]) == n       -> solucao unica
#   posto(A) == posto([A|b]) <  n       -> infinitas solucoes

def classifica(A, b):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float).reshape(-1, 1)
    n = A.shape[1]
    rA = np.linalg.matrix_rank(A)
    rAb = np.linalg.matrix_rank(np.hstack([A, b]))
    if rA != rAb:
        tipo = "nenhuma solucao"
    elif rA == n:
        tipo = "solucao unica -> " + str(np.linalg.solve(A, b.flatten()))
    else:
        tipo = "infinitas solucoes"
    return rA, rAb, tipo

sistemas = {
    "a": ([[1, 1], [2, 2]], [3, 6]),
    "b": ([[1, 1], [2, 2]], [3, 7]),
    "c": ([[1, 1], [2, -1]], [3, 0]),
}

for nome, (A, b) in sistemas.items():
    rA, rAb, tipo = classifica(A, b)
    print(f"3{nome}) posto(A) = {rA}, posto([A|b]) = {rAb}  =>  {tipo}")
```

## Exercício 4 — Eliminação de Gauss sem pivoteamento

O programa recebe $A$ e $b$, monta a matriz aumentada, zera os elementos abaixo de cada pivô, faz a substituição regressiva e mostra a solução.

```python
def gauss_sem_pivoteamento(A, b, mostrar=True):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # monta a matriz aumentada [A | b]
    M = np.hstack([A, b.reshape(-1, 1)])
    if mostrar:
        print("Matriz aumentada inicial:\n", M)

    # eliminacao: pra cada pivo k, zera os elementos abaixo dele
    for k in range(n - 1):
        if M[k, k] == 0:
            raise ValueError(f"Pivo zero na linha {k+1}, precisaria de pivoteamento")
        for i in range(k + 1, n):
            m = M[i, k] / M[k, k]               # multiplicador
            M[i, k:] = M[i, k:] - m * M[k, k:]  # L_i <- L_i - m*L_k
        if mostrar:
            print(f"\nDepois de zerar a coluna {k+1}:\n", M)

    # substituicao regressiva (de baixo pra cima)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.dot(M[i, i+1:n], x[i+1:n])
        x[i] = (M[i, n] - soma) / M[i, i]

    return x
```

```python
# teste com o sistema do exercicio 2
A2 = [[ 2,  1, -1],
      [-3, -1,  2],
      [-2,  1,  2]]
b2 = [8, -11, -3]

x_gauss = gauss_sem_pivoteamento(A2, b2)
x_numpy = np.linalg.solve(A2, b2)

print("\nSolucao (meu Gauss):", x_gauss)
print("Solucao (numpy):    ", x_numpy)
print("Diferenca maxima:   ", np.max(np.abs(x_gauss - x_numpy)))
```

**Interpretação:** o resultado deu $x = 2,\ y = 3,\ z = -1$, igual à resolução manual do exercício 2 e ao `numpy.linalg.solve`. As matrizes intermediárias também são as mesmas que foram feitas à mão. Se aparecer alguma diferença entre os dois métodos, ela é da ordem de $10^{-15}$, que é só erro de arredondamento de ponto flutuante.

Uma limitação do método sem pivoteamento é que ele para se aparecer um pivô zero, e pode ficar instável numericamente se o pivô for muito pequeno. Por isso o `numpy` usa fatoração LU com pivoteamento parcial.

## Exercício 5 — Modelo de balanço térmico

### a) Resolução manual

$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
-1 & 4 & -1 & 10 \\
0 & -1 & 3 & 10
\end{array}\right]
$$

**Etapa 1:** $L_2 \leftarrow L_2 + \frac{1}{4}L_1$ (a linha 3 já tem zero na coluna 1)

$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
0 & 15/4 & -1 & 55/4 \\
0 & -1 & 3 & 10
\end{array}\right]
$$

**Etapa 2:** $L_3 \leftarrow L_3 + \frac{4}{15}L_2$

- coluna 3: $3 - \frac{4}{15} = \frac{41}{15}$
- termo independente: $10 + \frac{4}{15}\cdot\frac{55}{4} = 10 + \frac{11}{3} = \frac{41}{3}$

$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
0 & 15/4 & -1 & 55/4 \\
0 & 0 & 41/15 & 41/3
\end{array}\right]
$$

**Substituição regressiva:**

- $T_3 = \dfrac{41/3}{41/15} = \dfrac{15}{3} = 5$
- $\frac{15}{4}T_2 - 5 = \frac{55}{4} \Rightarrow \frac{15}{4}T_2 = \frac{75}{4} \Rightarrow T_2 = 5$
- $4T_1 - 5 = 15 \Rightarrow T_1 = 5$

**Solução:** $T_1 = T_2 = T_3 = 5$

Verificação na 3ª equação: $-5 + 3(5) = 10$ ✓

### b) Implementação em Python

```python
# Exercicio 5b - reaproveitando a funcao do exercicio 4
A5 = [[ 4, -1,  0],
      [-1,  4, -1],
      [ 0, -1,  3]]
b5 = [15, 10, 10]

T = gauss_sem_pivoteamento(A5, b5)
print("\nTemperaturas (Gauss):", T)
print("Temperaturas (numpy):", np.linalg.solve(A5, b5))

# residuo pra conferir: A*T - b tem que dar ~0
print("Residuo A*T - b:     ", np.array(A5) @ T - np.array(b5))
```

```python
# conferindo se a matriz eh diagonalmente dominante
# (|a_ii| > soma dos outros |a_ij| da linha)
A5 = np.array(A5, dtype=float)
for i in range(len(A5)):
    diag = abs(A5[i, i])
    resto = np.sum(np.abs(A5[i])) - diag
    print(f"linha {i+1}: |a_ii| = {diag:.0f}  >  {resto:.0f} ?  {diag > resto}")
```

**Interpretação:** as três temperaturas deram iguais ($T_1 = T_2 = T_3 = 5$), ou seja, com essas condições a rede fica em equilíbrio com temperatura uniforme. O resultado do código bate com a resolução manual e com o `numpy`, e o resíduo deu zero.

A matriz é estritamente diagonalmente dominante (em cada linha o elemento da diagonal é maior que a soma dos outros em módulo), então a eliminação sem pivoteamento funciona bem aqui, sem pivôs nulos nem problemas de estabilidade.
